# BraTS 2021 nnU-Net (DynUNet) Baseline — Full 65 Epochs Runner on Kaggle

Notebook chuẩn hóa chạy **nnU-Net (DynUNet) Baseline trên TOÀN BỘ tập dữ liệu BraTS 2021 (1,251 ca: ~1,000 Train / ~125 Val / ~126 Held-Out Test) đủ 65 Epochs**.

### Đặc tả cấu hình nnU-Net Baseline 65 Epochs:
- **Kiến trúc mạng**: **`DynUNet` (MONAI)** với chuẩn hóa **`InstanceNorm3d`** và kích hoạt **`LeakyReLU(0.01)`**.
- **Độ chính xác**: Thuần **`FP32` (`--no-amp`)** cho sự ổn định số học tối đa trên GPU NVIDIA T4.
- **Phần cứng**: **`GPU T4 x2`** (`torch.nn.DataParallel` tự động nhận diện cả 2 GPU) + 4 CPU workers (`persistent_workers=True`).
- **Lịch trình huấn luyện**: **`65` epochs**, 5 mốc Validation (`val_interval=13`: epochs 13, 26, 39, 52, 65). Tự động lưu checkpoint tốt nhất theo Validation Mean Dice.
- **Dữ liệu**: Tự động giải nén `BraTS2021_Training_Data.tar` vào `/kaggle/tmp` (ổ SSD tạm), tiêu tốn **0 MB** dung lượng `/kaggle/working`.
- **Đánh giá Held-Out Test**: Tự động đánh giá độc lập trên toàn bộ ~126 ca tập Test (10%) sau khi hoàn tất 65 epochs.
- **Thời gian ước tính**: ~7.5 giờ trên cụm 2x GPU T4 (hoàn toàn an toàn trong hạn mức 12h của Kaggle).

```python
# 3. Huấn luyện nnU-Net Baseline trên TOÀN BỘ 1,251 ca BraTS 2021 (Tiếp nối lên 65 Epochs)
#    - Kiểm tra nạp checkpoint 50 epochs (nếu có trong /kaggle/input) để train nối tiếp từ Epoch 51..65 (chỉ mất ~1.6h!)
#    - DynUNet với InstanceNorm3d + LeakyReLU
#    - DataParallel 2x T4 GPU, batch size 4, thuần FP32 (--no-amp)
#    - Sau khi hoàn thành 65 epochs, tự động đánh giá Held-Out Test trên toàn bộ ~126 ca test
from pathlib import Path
import subprocess

resume_ckpt = None
for p in Path("/kaggle/input").rglob("latest_model.pth"):
    resume_ckpt = str(p)
    break

cmd = [
    "python", "-u", "/kaggle/tmp/repo/scripts/run_nnunet_baseline.py",
    "--epochs", "65",
    "--val-interval", "13",
    "--roi-size", "96",
    "--batch-size", "4",
    "--sw-batch-size", "4",
    "--lr", "1e-3",
    "--optimizer", "adamw",
    "--loss-type", "dice_bce",
    "--overlap", "0.5",
    "--cache-rate", "0.0",
    "--num-workers", "4",
    "--no-amp",
    "--keep-latest",
    "--max-hours", "11.5",
    "--output-dir", "/kaggle/working/outputs/nnunet",
    "--checkpoint-dir", "/kaggle/working/checkpoints/nnunet",
]
if resume_ckpt:
    print(f"[Resume] Nạp checkpoint 50 epochs: {resume_ckpt} -> Tiếp tục huấn luyện từ Epoch 51..65 (~1.6h)!")
    cmd.extend(["--resume", resume_ckpt])
else:
    print("[Start] Chạy từ Epoch 1..65")

subprocess.run(cmd, check=True)
```

```python
# 4. Đọc và trực quan hóa toàn bộ kết quả thực nghiệm Held-Out Test & Biểu đồ huấn luyện
import json
from IPython.display import Image, display

report_path = "/kaggle/working/outputs/nnunet/nnunet_metrics_report.json"
try:
    with open(report_path, "r", encoding="utf-8") as f:
        rep = json.load(f)

    print(f"=== KẾT QUẢ THỰC NGHIỆM nnU-Net (DynUNet) 65 EPOCHS TRÊN TOÀN BỘ BraTS 2021 ===")
    print(f"Tổng số ca bệnh nhân: {rep['dataset_split']['total_cases']}")
    print(f"Phân chia tập: Train={rep['dataset_split']['train_cases']}, Val={rep['dataset_split']['val_cases']}, Test={rep['dataset_split']['test_cases']}")
    print(f"Epoch Val tốt nhất: {rep['best_val_epoch']} | Best Val Mean Dice: {rep['best_val_dice_mean']:.4f}")
    print("\n--- TỔNG HỢP KẾT QUẢ TẬP HELD-OUT TEST (MEAN) ---")
    print(json.dumps(rep["test_summary_mean"], indent=2))
    print("\n--- PHÂN PHỐI THỐNG KÊ TẬP TEST (MEAN / STD / MEDIAN) ---")
    print(json.dumps(rep["test_distribution_stats"], indent=2))

    display(Image(filename="/kaggle/working/outputs/nnunet/nnunet_learning_curves.png"))
    display(Image(filename="/kaggle/working/outputs/nnunet/nnunet_test_predictions.png"))
except Exception as e:
    print("Lỗi hiển thị kết quả:", e)
```
